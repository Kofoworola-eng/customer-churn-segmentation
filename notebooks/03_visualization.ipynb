```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

rfm = pd.read_csv('../data/processed/rfm_segmented.csv')
rfm.head()
```

```python
plt.figure(figsize=(10, 6))
segment_order = rfm['Segment'].value_counts().index

sns.countplot(data=rfm, x='Segment', order=segment_order, palette='viridis')
plt.title('Customer Distribution by RFM Segment', fontsize=14)
plt.xlabel('Segment')
plt.ylabel('Number of Customers')
plt.xticks(rotation=30)
plt.tight_layout()
plt.savefig('../dashboard/segment_distribution.png', dpi=150)
plt.show()
```
